DATA VISUALIZATION



```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure()
plt.hist(df['G3'])
plt.xlabel("Final Grade (G3)")
plt.ylabel("Number of Students")
plt.title("Distribution of Final Grades")
plt.show()
```

```python
plt.figure()
sns.barplot(x='sex', y='G3', data=df)
plt.title("Gender vs Final Grade")
plt.show()
```

```python
plt.figure()
sns.heatmap(df.corr(numeric_only=True), cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

```python
### Conclusion
# Dataset had no missing values.
# Students with higher study time scored better.
# Final grades are positively correlated with previous grades.
# Gender has slight impact on performance.
```
